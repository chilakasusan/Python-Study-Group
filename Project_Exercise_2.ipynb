# Exercise 4: Working with Data Types


```python
# Create one variable for each of the following data types:
#Integer
#Float
#String
#Boolean

integer_value = 100
float_value = 3.234
string_value = "Python"
boolean_value = True

# Display the data types using type()
print(type(integer_value))
print(type(float_value))
print(type(string_value))
print(type(boolean_value))
```

# Exercise 5: Dynamic Typing

```python
#Create a variable with an integer value.
integer_value = 100

#Print its data type.
print("\nBefore Change:")
print(type(integer_value))

#Change the variable to a string.
integer_value = "Python"

# Print the new data type
print("\nAfter Change:")
print(type(integer_value))


```

# Mini Project: Student Information Card


```python
#Project Scenario A training institute wants to register students for a Python course. 
#Your task is to create a simple program that collects student information and displays it in a professional format.


# Collect information from the student
name = input("Enter your name: ")
age = input("Enter your age: ")
course = input("Enter your course: ")
country = input("Enter your country: ")

# Display Student Information Card
print("==============================")
print("     STUDENT INFORMATION CARD")
print("==============================")

print("\nName    :", name)
print("Age     :", age)
print("Course  :", course)
print("Country :", country)

# Display data types
print("\n------------------------------")
print("Data Types")
print("------------------------------")

print("Name    :", type(name))
print("Age     :", type(age))
print("Course  :", type(course))
print("Country :", type(country))
```

```python
#Bonus Challenge

#Converting the user's age into an integer using int().

age = int(age)

print("\n------------------------------")
print("After Conversion")
print("------------------------------")

#Displaying the new data type.

print("Age:", age)
print("\nData Type:")
print(type(age))

# Printing a confirmation message showing that the conversion was successful.
print("\nAge conversion was successful!")
```
