Hands-On 1: Student Score Tracker

```python
score = 60
print("Starting score:", score)

score += 15   # add 15 marks for assignment
print("After assignment:", score)

score -= 5    # subtract 5 marks for late submission
print("After late penalty:", score)

score *= 2    # double the score as bonus
print("Final score:", score)

#Bonus
score = int(input("Enter the initial score: "))

score += 15
score -= 5
score *= 2

print("Final score:", score)
```

Hands-On 2: Digital Wallet Simulator

```python
balance = 25000
print("Starting balance: ₦", balance)

balance += 8500   # deposit
print("After deposit: ₦", balance)

balance -= 4200   # withdraw
print("After withdrawal: ₦", balance)

balance /= 3      # split across 3 savings accounts
print("Final balance per account: ₦", balance)

print("Data type of balance:", type(balance)) #data type

#bonus

balance = 50000

deposit = float(input("Enter deposit amount: "))
balance += deposit

withdrawal = float(input("Enter withdrawal amount: "))
balance -= withdrawal

balance /= 3
print("Final balance per account: ₦", balance)

print("Data type of balance:", type(balance)) #data type
```

Hands-On 3: Inventory Management System (Mini Project)

```python
stock = 150
print("Starting stock:", stock)

stock -= 28   # customer buys 28 items
print("After sale:", stock)

stock += 75   # new shipment arrives
print("After restock:", stock)

stock //= 12  # pack into cartons of 12
print("Number of full cartons:", stock)
```

Bonus Challenge

```python
stock = int(input("Enter initial stock: "))

sold = int(input("How many items were sold? "))
stock -= sold
print("Remaining stock after sales:", stock)

new_items = int(input("How many new items arrived? "))
stock += new_items
print("Updated stock after restocking:", stock)

carton_size = int(input("How many items fit in one carton? "))

# Save a copy before modifying, so we can calculate leftovers correctly
full_cartons = stock // carton_size
leftover = stock % carton_size

print("Number of full cartons:", full_cartons)
print("Remaining loose items:", leftover)
```
