EXERCISE 1: Student Report Card

```python
student_name = "Esther Johnson"
age = 17
subject = "Mathematics"
score = 91.75
 
# A student passes if their score is 50 or above.
pass_message = f"Congratulations {student_name}!\nYou passed {subject}." \
    if score >= 50 else f"Keep practicing, {student_name}. You need to retake {subject}."
 
report_card = f"""==============================
        STUDENT REPORT
==============================
 
Student : {student_name}
Age     : {age} years
Subject : {subject}
Score   : {score}%
 
{pass_message}
 
=============================="""
 
print(report_card)
 
print("\n")
```

EXERCISE 2: Payroll System

```python
employee_name = "David Musa"
department = "Finance"
basic_salary = 185000
bonus = 35000
tax_rate = 7.5  # percentage
 
# --- Calculations ---
gross_salary = basic_salary + bonus
tax_amount = gross_salary * (tax_rate / 100)
net_salary = gross_salary - tax_amount
 
payroll_receipt = f"""==================================
        PAYROLL SUMMARY
==================================
 
Employee   : {employee_name}
Department : {department}
Basic Salary : ₦{basic_salary:,.2f}
Bonus        : ₦{bonus:,.2f}
Gross Salary  : ₦{gross_salary:,.2f}
Tax ({tax_rate}%)  : ₦{tax_amount:,.2f}
Net Salary : ₦{net_salary:,.2f}
 
=================================="""
 
print(payroll_receipt)
 
print("\n")
```

EXERCISE 3: E-commerce Order Receipt

```python

customer_name = "Amina Bello"
product = "Wireless Headphones"
quantity = 3
unit_price = 27500
 
# --- Calculations ---
total_cost = quantity * unit_price
 
DISCOUNT_RATE = 0.10  # 10% discount
qualifies_for_discount = quantity > 2
discount = total_cost * DISCOUNT_RATE if qualifies_for_discount else 0
final_amount = total_cost - discount
 
# Build the discount section only if a discount applies.
discount_lines = (
    f"Discount    : ₦{discount:,.2f}\n\nFinal Amount : ₦{final_amount:,.2f}\n\n"
    if qualifies_for_discount else ""
)
 
order_receipt = f"""====================================
      ONLINE STORE RECEIPT
====================================
 
Customer   : {customer_name}
Product    : {product}
Quantity   : {quantity}
Unit Price : ₦{unit_price:,.2f}
 
Total Cost : ₦{total_cost:,.2f}
 
{discount_lines}Thank you for shopping with us!
 
===================================="""
 
print(order_receipt)
 
print("\n")
 
 
# =========================================================
```
