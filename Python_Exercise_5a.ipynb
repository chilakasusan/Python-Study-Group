CASE STUDY PROJECT: Customer Feedback Management System

```python

# Task: Store customer feedback as strings and print a professionally formatted report using a multiline string.
 
# Stored feedback data 
review_customer_name = "Sarah Johnson"
product_name = "Wireless Mouse"
product_category = "Computer Accessories"
customer_rating = "5 Stars"
 
# Bonus: quotation marks inside the review, handled with a quote style that avoids unnecessary escaping.
customer_review = (
    'The product arrived on time, was easy to install, and exceeded '
    'my expectations.As Sarah put it, "I would definitely recommend '
    'it to others."'
)
 
company_name = "Insight Retail Technologies"
support_email = "support@insightretailtech.com"
 
welcome_message = """Welcome to the Insight Retail Technologies
Customer Feedback Management System.
Generating your feedback report now...
"""
 
feedback_report = f"""{'=' * 40}
       CUSTOMER FEEDBACK REPORT
{'=' * 40}
 
Customer Name : {review_customer_name}
Product       : {product_name}
Category      : {product_category}
Rating        : {customer_rating}
 
Customer Review
{'-' * 40}
{customer_review}
 
{'=' * 40}
Thank you for your feedback!
Need help? Contact us at {support_email}
{company_name}
{'=' * 40}"""
 
print(welcome_message)
print(feedback_report)
 
```
