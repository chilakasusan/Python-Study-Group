Question 1 — Employee Payroll & Bonus Eligibility

```python
#A company pays its employees a monthly salary and a performance bonus.
#Given:
#Basic Salary = ₦180,000
#Performance Bonus = ₦45,000
#Tasks
#Calculate the employee's total salary.

#let basic salary = X
#let performance bonus = Y

X = 180000
Y = 45000

Total_Salary  = X + Y

#Display Result
print("Total Salary: ₦{:,.0f}".format(Total_Salary))

#Determine whether the employee qualifies for a premium benefits package.
#Eligibility Rule
#An employee qualifies if the total salary is greater than or equal to ₦220,000.


Eligible_for_premium = Total_Salary >= 220000

# Display the results

print("Eligible for Premium Benefits:", Eligible_for_premium)
```

Question 2 — Student Performance Evaluation

```python
# Student Performance Evaluation

# Scores
python_score = 72
sql_score = 65
excel_score = 80
powerbi_score = 75

# Calculate the total score
total_score = python_score + sql_score + excel_score + powerbi_score
print("Total Score:", total_score)

# Calculate the average score
average_score = total_score / 4
print("Average Score:", average_score)

# Determine if student passed
#A student passes if the average score is greater than or equal to 70.
passed = average_score >= 70
print("Passed:", passed)



```

Question 3 — Inventory Management System

```python
#Stock available at warehouse
Initial_Stock = 250
#Stock sold during the day:
Sold = 85

#Calculate the remaining stock

Remaining_Stock  = Initial_Stock - Sold

#Display Result
print("Remaining Stock:", Remaining_Stock, "units")
      
#Determine if reorder is needed
reorder_needed = Remaining_Stock < 200     

#Display Result
print("Reorder Needed:", reorder_needed)
```

Question 4 — Banking Withdrawal Checker

```python
# Account balance
Account_Balance = 520000

# Withdrawal amount requested
Withdrawal_Amount = 150000

# Calculate remaining balance 
Remaining_Balance = Account_Balance - Withdrawal_Amount


# Determine whether the withdrawal can be approved
Withdrawal_Approved = Account_Balance >= Withdrawal_Amount


# Display Results

print("Remaining Balance: ₦{:,.0f}".format(Remaining_Balance))

print("Withdrawal Approved:", Withdrawal_Approved)
```

Bonus Challenge — Smart Shopping Discount Calculator

```python
# Item prices
Laptop_Bag = 18500
Wireless_Mouse = 12000
Keyboard = 24500
USB_Drive = 5000

#Calculate the total purchase amount

Total_Purchase = Laptop_Bag + Wireless_Mouse + Keyboard + USB_Drive

print("Total Purchase: ₦{:,.0f}".format(Total_Purchase))

# Determine discount qualification (business rule: >= ₦50,000 gets 10% off)
Discount_Qualified = Total_Purchase >= 50000

print("Discount Qualified:", Discount_Qualified)

# Calculate discount amount (10% if qualified, otherwise 0)
if Discount_Qualified:
    Discount_Amount = Total_Purchase * 0.10
else:
    Discount_Amount = 0
        
print("Discount Amount: ₦{:,.0f}".format(int(Discount_Amount)))


# Calculate final amount to pay
Final_Amount = Total_Purchase - Discount_Amount

print("Final Amount: ₦{:,.0f}".format(int(Final_Amount)))
```
